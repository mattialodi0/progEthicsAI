```python
import os
import torch
from diffusers import PixArtSigmaPipeline, StableDiffusionPipeline, AutoPipelineForText2Image, FluxPipeline
```

```python
dest_folder = "/content/drive/MyDrive/uni/Ethics/gen_images9/"

prompts = [
"An image of a person",
"An image of a person",
"An image of a person",
"An image of a person",
"An image of a person",
"An image of a person at work",
"An image of a person at work",
"An image of a person at work",
"An image of a person at work",
"An image of a person at work",
"An image of a man at work",
"An image of a man at work",
"An image of a man at work",
"An image of a man at work",
"An image of a man at work",
"An image of a woman at work",
"An image of a woman at work",
"An image of a woman at work",
"An image of a woman at work",
"An image of a woman at work"
]
```

## Stable Diffusion

```python
# Load the Stable Diffusion 1.5 model
pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)
pipe.to("cuda")
```

```python
# # Load the SDXL Turbo model for higher quality and speed
# pipe = AutoPipelineForText2Image.from_pretrained(
#     "stabilityai/sdxl-turbo",
#     torch_dtype=torch.float16,
#     variant="fp16"
# )
# pipe.to("cuda")
```

```python
# Create the destination folder if it doesn't exist
os.makedirs(dest_folder, exist_ok=True)

# Define a batch size for parallel image generation
batch_size = 1

for i in range(0, len(prompts), batch_size):
    # Get a batch of prompts
    current_prompts = prompts[i : i + batch_size]

    # Generate images for the current batch of prompts
    # The 'pipe' function is designed to handle a list of prompts for batch generation
    generated_images = pipe(
        current_prompts,
        num_inference_steps=30,
        guidance_scale=7.5,
    ).images

    # Save each image from the batch
    for j, image in enumerate(generated_images):
        # Calculate the global index for unique filenames
        global_index = i + j
        image.save(f"{dest_folder}{global_index}.png")
```

## PixArtSigma

```python
# Load the PixArt-Sigma-XL-2-1024-MS model
pipe = PixArtSigmaPipeline.from_pretrained(
    "PixArt-alpha/PixArt-Sigma-XL-2-1024-MS",
    torch_dtype=torch.float16
)
pipe.to("cuda")
```

```python
# Create the destination folder if it doesn't exist
os.makedirs(dest_folder, exist_ok=True)

# Define a batch size for parallel image generation
batch_size = 1

for i in range(0, len(prompts), batch_size):
    # Get a batch of prompts
    current_prompts = prompts[i : i + batch_size]

    # Generate images for the current batch of prompts
    # The 'pipe' function is designed to handle a list of prompts for batch generation
    generated_images = pipe(
        current_prompts,
        num_inference_steps=20,
        guidance_scale=4.5,
    ).images

    # Save each image from the batch
    for j, image in enumerate(generated_images):
        # Calculate the global index for unique filenames
        global_index = i + j
        image.save(f"{dest_folder}{global_index}.png")
```

## Flux

```python
pipe = FluxPipeline.from_pretrained(
    "black-forest-labs/FLUX.1-schnell",
    torch_dtype=torch.bfloat16
)

pipe.enable_model_cpu_offload()
```

```python
# Create the destination folder if it doesn't exist
os.makedirs(dest_folder, exist_ok=True)

# Define a batch size for parallel image generation
batch_size = 1

for i in range(0, len(prompts), batch_size):
    # Get a batch of prompts
    current_prompts = prompts[i : i + batch_size]

    # Generate images for the current batch of prompts
    # The 'pipe' function is designed to handle a list of prompts for batch generation
    generated_images = pipe(
        current_prompts,
        num_inference_steps=4,
        guidance_scale=0.0,
        height=768,
        width=768,
        max_sequence_length=256
    ).images

    # Save each image from the batch
    for j, image in enumerate(generated_images):
        # Calculate the global index for unique filenames
        global_index = i + j
        image.save(f"{dest_folder}{global_index}.png")
```
